# Lab Exercise – Payroll Tax Functions

This lab demonstrates how to create and use custom Python functions to calculate payroll taxes.

The notebook includes:
- A function to calculate Social Security tax
- A function to calculate Medicare tax
- A function to calculate federal withholding tax using conditional statements
- Test cases using different gross pay amounts and withholding codes

This exercise practices:
- Function creation
- Parameters and return values
- Mathematical calculations
- if / else conditional logic

```python
# Social Security Tax Function
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
    
```

```python
# Medicare Tax Function
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23


    elif withholding_code == 1:
        tax_rate = 0.21

    elif withholding_code == 2:
        tax_rate = 0.195

    elif withholding_code == 3:
        tax_rate = 0.185

    else:
        tax_rate = 0.18 - ((withholding_code - 4) * 0.005)

    return gross_pay * tax_rate


    
    
```

```python
gross_pay = 750
withholding_code = 0

print("Person 1")
print(f"Social Security Tax: ${get_soc_sec_tax(gross_pay):.2f}")
print(f"Medicare Tax: ${get_medicare_tax(gross_pay):.2f}")
print(f"Federal Tax: ${get_federal_tax(gross_pay, withholding_code):.2f}")
```
