```python
def favorite_things(name, movie, music):
    return(
        f"My name is {name}\n"
        f"My favorite movie is {movie}\n"
        f"My favorite music duo is {music}"
)
```

```python
print(favorite_things("Charles", "Shoalin Soccer", "$uicideboy$"))
```

```python
def why_im_here(purpose):
    return purpose
print(
    why_im_here("I joined Year Up United in order to find a stable career and be able to provide for my small family")
               )
    
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."

```

```python
print(favorite_place("Shenandoah National Park", "I driving along the moutain and stopping by to take in the beautiful view"))
```

```python
print(favorite_place("Sushi Queen", "I love trying all the difference sushi rolls they have"))
```
