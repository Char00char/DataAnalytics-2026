## Lab 2 - Exercise 2.B: Working with Datetime Functions


```python
# Import the datetime module
import datetime

```

# Today

```python
# Get the current date and time
today = datetime.datetime.now()
```

```python
# Format and print today's date
print(today.strftime("Today is %A, %B, %d, %Y"))
                     
```

# My Birthday

```python
birthday = datetime.date(1997, 8, 1)
```

```python
# Format and print birthday
print(birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
# Calculate the date 90 days from today
ninety_d = today + datetime.timedelta(days=90)
```

```python
#Format and print future date
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Time for Dinner

```python
# Define a dinner time
dinner_time = datetime.time(19, 30)
```

```python
# Format and print dinner time 
print(dinner_time.strftime("Let's meet for dinner at %I:%M %p"))
```
