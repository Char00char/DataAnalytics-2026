```python
import numpy as np
```

## Array Basics

```python
# Create a standard Python list
fruits = ["apple", "banana", "cherry", "date"]

print(fruits)
```

```python
#Create a one-dimensional NumPy array
arr_1 = np.array(fruits)

print(arr_1)
```

```python
# Check the data type of values in arr_1
print(arr_1.dtype)

# The returned dtype will likely be something like <U6
# U means Unicode string
# 6 means the maximum string length in the array is 6 characters
```

```python
# Create a two-dimensional array
arr_2 = np.array([
    [10, 11, 12],
    [13, 14, 15]
])

print(arr_2)
```

```python
# Print the transposed version of arr_2
print(arr_2.T)
```

```python
# Check whether arr_2 was permanently changed
print(arr_2)

# No, using .T does not permanently alter arr_2.
# It only displays a transposed view of the array.
```

```python
# Create another two-dimensional array
arr_3 = np.array([
    [1, 3, 5],
    [2, 4, 6]
])

print(arr_3)
```

```python
# Create arr_4 using slicing/indexing
arr_4 = np.array([
    arr_3[0, 2],
    arr_3[1, 1]
])

print(arr_4)
```

```python
# Create a three-dimensional array of all 1's
arr_5 = np.ones((2, 2, 2))

print(arr_5)
```
