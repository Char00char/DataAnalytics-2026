```python
# Function to get supplier code (before :)
def get_supplier_code(part_code):
    for i in range(len(part_code)
        if part_code[i] == ':':
            return part_code[:i]
                
```

```python
# Function to get product number (between : and -)
def get_product_num(part_code):
    colon_index = 0
    dash_index = 0

    for i in range(len(part_code)):
        if part_code[i] == ':':
            colon_index = i

        if part_code[i] == '-':
            dash_index = i

    return part_code[colon_index + 1:dash_index]

# Function to get size (after -)
def get_size(part_code):
    for i in range(len(part_code)):
        if part_code[i] == '-':
            return part_code[i + 1:]

# Example product codes
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'

# Test Example 1
print(f"For {part1}, the supplier is {get_supplier_code(part1)}.")
print(f"Part number {get_product_num(part1)] is avaiable in size {get_size(part1)}.")
print()


    
    

```
