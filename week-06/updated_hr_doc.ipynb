## Lab 3. Employee data
#This lab exercise makes updates to information extracted from an employee database

```python
# Create the hr_list variable

hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'), 
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
#Update Mark's last name to Blick-Hawley

hr_list[1] = (
    hr_list[1][0],
    hr_list[1][1],
    'Mark Blick-Hawley',
    hr_list[1][3]
)
```

```python
# Print updated list after Mark's change
print(hr_list)
```

```python
#Update Jim's department and salary

hr_list[3] = (
    hr_list[3][0],
    'CS',
    hr_list[3][2],
    '60000'
)
```

```python
#Print updated list after Jim's changes

print(hr_list)
```

```python
# Display formatted employee information
for employee in hr_list:
    emp_num = employee[0]
    dept = employee[1]
    name = employee[2]
    salary = int(employee[3])

    print(f"{emp_num} | {dept} | {name} | {salary:,}")
    
```
