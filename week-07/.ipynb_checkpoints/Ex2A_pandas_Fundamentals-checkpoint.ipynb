# Exercise 2.A Fundamentals of pandas

```python
import numpy as np
import pandas as pd
```

## Series

```python
# Plain Python objects

aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

```python
# Create Series objects

ser1 = pd.Series(aa)

ser2 = pd.Series(ab)

ser3 = pd.Series(list(ac))

ser4 = pd.Series(ad)

print(ser1)
print()
print(ser2)
print()
print(ser3)
print()
print(ser4)
```

```python
# Attempt to create ser5 incorrectly

ser5 = pd.Series(17, 18, 19, 20)

# TypeError:
# Series.__init__() takes from 1 to 4 positional arguments but 5 were given
```

## DataFrames from Series

```python
# Create DataFrame from ser1

df1 = pd.DataFrame(ser1)

df1
```

```python
# Add new columns using insert()

df1.insert(1, "Column2", ser2)

df1.insert(2, "Column3", ser3)

df1
```

```python
# Add third column using ser4

df1.insert(3, "Column4", ser4)

df1

# ser4 uses dictionary keys as labels/indexes instead of numeric indexes.
# Because the indexes do not match df1's indexes, pandas fills the rows with NaN values.
```

```python
# Add a new row using loc

df1.loc[4] = [5, 9, 12, 0]

df1
```

## Reading Data into a DataFrame

```python
# Read sakila_example.csv

df2 = pd.read_csv("sakila_example.csv")

df2
```

```python
# Display top 5 rows

df2.head()
```

```python
# Display last 10 rows

df2.tail(10)
```

```python
# Show summary information

df2.info()
```

A few facts we can determine from the info results:

- The number of rows and columns in the dataset
- The names of all columns
- The data type of each column
- Whether there are missing values
- Approximate memory usage of the DataFrame

```python
# Show statistics

df2.describe()
```

Useful insights from describe():

- Mean and average values for numeric columns
- Minimum and maximum values
- Standard deviation and spread of the data
- Quartiles that show data distribution

Less useful summarizations:

- Statistics on ID columns are usually not meaningful
because IDs are identifiers, not measurable quantities.

```python
df3 = pd.read_csv("northwind_example.csv")

df3
```

```python
# Describe df3

df3.describe()
```

Useful insights from df3 describe():

- Shows averages and ranges for numeric business data
- Helps identify large or small values
- Helps spot unusual outliers

The summarizations may differ from df2 because this dataset
likely contains different types of columns and business-related values.
Different datasets produce different statistical summaries
depending on the data types included.

## Coding a DataFrame


```python
# Create dictionary

test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Create DataFrame

df4 = pd.DataFrame(test_data)

df4
```

```python
# Export DataFrame to CSV

df4.to_csv("test_data.csv", index=False)
```
