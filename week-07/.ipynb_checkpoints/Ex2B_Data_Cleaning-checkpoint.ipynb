# Exercise 2.B Cleaning up data

```python
import numpy as np
import pandas as pd
```

```python
# Read test_data.csv into a DataFrame

df = pd.read_csv("test_data.csv")

df
```

```python
# Show column statistics

df.describe()
```

## Filtering a DataFrame

```python
# Filter rows where Age is greater than 18

df[df["Age"] > 18]
```

```python
# Filter rows where Score is greater than 50 and less than 100

df[(df["Score"] > 50) & (df["Score"] < 100)]
```

## Modifying a DataFrame

```python
# Remove rows where Score is less than 0

df_cleaned = df.drop(df[df["Score"] < 0].index)

df_cleaned
```

```python
# Remove the City column

df_cleaned = df_cleaned.drop(columns=["City"])

df_cleaned
```

```python
# Replace missing Age values with the mean Age

mean_age = df_cleaned["Age"].mean()

df_cleaned["Age"] = df_cleaned["Age"].fillna(mean_age)

df_cleaned
```

```python
# Replace Score value of 0 with the mean Score

mean_score = df_cleaned["Score"].mean()

df_cleaned["Score"] = df_cleaned["Score"].replace(0, mean_score)

df_cleaned
```

```python
# Remove remaining rows with NaN values

df_cleaned = df_cleaned.dropna()

df_cleaned
```

```python
# Set ID as index labels and remove the ID column

df_cleaned = df_cleaned.set_index("ID")

df_cleaned
```
