# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
# Create arrA
arrA = np.array([[11, 12, 13],
                 [14, 15, 16]])

print(arrA)
```

```python
# Loop through the first dimension
for x in arrA:
    print(x * 2)
```

```python
# Loop through the second dimension
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row of the array, so multiplying by 2 affects an entire row at once.  
The second loop goes through each individual value inside the rows, so it prints each number separately multiplied by 2.

```python
# Loop through the second dimension again
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because x represents the entire row,
# not the individual value y. So the whole row is multiplied by 2 each time.
```

## Joining and Stacking Arrays

```python
# Concatenate on axis=0
joined_array_1 = np.concatenate((arrA, arrA))

print(joined_array_1)
```

```python
# Concatenate on axis=1
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
# Stack joined_array_1 with itself
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
# Attempt to stack arrays of different shapes
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# Error explanation:
# np.stack() requires arrays to have the exact same shape.
# joined_array_1 and joined_array_2 have different dimensions,
# so NumPy returns a shape mismatch error.
```

```python
# Concatenate joined_array_1 with arrA
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
# Attempt concatenation on axis=1
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# Error explanation:
# The arrays do not have the same number of rows,
# so NumPy cannot join them side-by-side on axis=1.
```

```python
# hstack joins arrays horizontally (side-by-side)
horizontal_stack = np.hstack((arrA, arrA))
print(horizontal_stack)

# vstack joins arrays vertically (top-to-bottom)
vertical_stack = np.vstack((arrA, arrA))
print(vertical_stack)

# dstack joins arrays depth-wise (creates a third dimension)
depth_stack = np.dstack((arrA, arrA))
print(depth_stack)
```

## Splitting Arrays

```python
# Split joined_array_1 into 2 sections
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# Result explanation:
# This creates 2 arrays.
# Each array is still 2-dimensional.
```

```python
# Split joined_array_1 into 3 sections
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# Result explanation:
# This creates 3 arrays.
# Each resulting array is still 2-dimensional,
# but the rows are divided as evenly as possible.
```

## Searching and Replacing Array Values

```python
# Replace odd numbers with 'odd'
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)
```

## Sorting and Filtering Arrays

```python
# Create arrB
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

# Sort arrB
arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
# Filter values less than 7
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
# Filter values greater than or equal to 7
y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
