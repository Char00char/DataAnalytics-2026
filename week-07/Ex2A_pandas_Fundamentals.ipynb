```python
import numpy as np
import pandas as pd
```

## Series

```python
# Plain Python objects

aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

```python
# Create Series objects

ser1 = pd.Series(aa)

ser2 = pd.Series(ab)

ser3 = pd.Series(list(ac))

ser4 = pd.Series(ad)

print(ser1)
print()
print(ser2)
print()
print(ser3)
print()
print(ser4)
```

```python
# Attempt to create ser5 incorrectly

ser5 = pd.Series(17, 18, 19, 20)

# TypeError:
# Series.__init__() takes from 1 to 4 positional arguments but 5 were given
```

## DataFrames from Series

```python
# Create DataFrame from ser1

df1 = pd.DataFrame(ser1)

df1
```

```python
# Add new columns using insert()

df1.insert(1, "Column2", ser2)

df1.insert(2, "Column3", ser3)

df1
```

```python
# Add third column using ser4

df1.insert(3, "Column4", ser4)

df1

# ser4 uses dictionary keys as labels/indexes instead of numeric indexes.
# Because the indexes do not match df1's indexes, pandas fills the rows with NaN values.
```

```python
# Add a new row using loc

df1.loc[4] = [5, 9, 12, 0]

df1
```
