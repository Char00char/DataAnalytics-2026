# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
# Random integers between 1 and 20 with 2 elements
r_1 = np.random.randint(1, 21, size=2)

print(r_1)
```

```python
# Random integers between 1 and 20 with 1 row and 2 columns
r_2 = np.random.randint(1, 21, size=(1, 2))

print(r_2)
```

```python
# Random floats between 0 and 1 with 2 rows and 1 column
r_3 = np.random.rand(2, 1)

print(r_3)
```

```python
# Random integers between 20 and 100 with 4 rows and 5 columns
r_4 = np.random.randint(20, 101, size=(4, 5))

print(r_4)
```

```python
# Random floats between 0 and 1 with 4 rows and 5 columns
r_5 = np.random.rand(4, 5)

print(r_5)
```

## Generating Random Choices from an Array

```python
# Array of student names
students = np.array([
    "Alex", "Brianna", "Carlos", "Danielle",
    "Ethan", "Fatima", "Grace", "Hector",
    "Isabella", "Jordan", "Kayla", "Liam"
])

print(students)
```

```python
# Choose one random student
call_on = np.random.choice(students)

print(call_on)
```

```python
# Random breakout groups
breakout_groups = np.random.choice(students, size=(3, 4))

print(breakout_groups)

# This may not be the best way to assign breakout groups
# because the same student can appear multiple times
# while some students may not appear at all.
```

```python
# Weighted random selections
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[
        0.30,  # Alex
        0.10,  # Brianna
        0.05,  # Carlos
        0.00,  # Danielle
        0.10,  # Ethan
        0.05,  # Fatima
        0.05,  # Grace
        0.10,  # Hector
        0.05,  # Isabella
        0.05,  # Jordan
        0.10,  # Kayla
        0.05   # Liam
    ]
)

print(call_on_weighted)

# Probability explanation:
# Alex has a 30% chance of being selected.
# Danielle has a 0% chance of being selected.
# The remaining students have smaller probabilities
# distributed across the rest of the class.
```

## Generating a Normal Distribution

```python
# Generate a normal distribution
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))

print(bell_curve)
```
