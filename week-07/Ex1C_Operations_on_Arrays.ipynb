## Sample Arrays

## Permutations

```python
# Print current r_4
print(r_4)
```

```python
# Shuffle r_4
np.random.shuffle(r_4)

print(r_4)
```

The shuffle function rearranges the rows of the array.  
The values inside each row stay together, but the order of the rows changes.

```python
# Create a permutation of r_4
new_shuffle = np.random.permutation(r_4)

print(new_shuffle)

# Print original r_4 again
print(r_4)
```

shuffle changes the original array directly.  
permutation creates a shuffled copy and leaves the original array unchanged.

## Mathematical Operations

```python
# Add r_1 and r_2

print(r_1)
print(r_1.shape)

print(r_2)
print(r_2.shape)

result_1 = np.add(r_1, r_2)

print(result_1)
print(result_1.shape)
```

```python
# Add r_1 and r_3

print(r_1)
print(r_1.shape)

print(r_3)
print(r_3.shape)

result_2 = np.add(r_1, r_3)

print(result_2)
print(result_2.shape)
```

```python
# Add r_2 and r_4
r_2 = np.random.randint(1, 21, size=(1,5))

print(r_2)
print(r_2.shape)

print(r_4)
print(r_4.shape)

result_3 = np.add(r_2, r_4)

print(result_3)
print(result_3.shape)
```

```python
# Attempt to add r_1 and r_4

result_4 = np.add(r_1, r_4)

print(result_4)

# Error explanation:
# r_1 has only 2 values while r_4 has 5 columns.
# Their shapes are incompatible for broadcasting.
```

```python
# Add np.array([1]) to r_4

print(np.add(r_4, np.array([1])))
```

```python
# Add np.array([[1]]) to r_4

print(np.add(r_4, np.array([[1]])))
```

```python
# Add np.array([1,1,1,1,1]) to r_4

print(np.add(r_4, np.array([1,1,1,1,1])))
```

```python
# Attempt to add [[1],[1]] to r_4

print(np.add(r_4, np.array([[1],[1]])))

# Error explanation:
# The array has shape (2,1) which does not match
# the shape of r_4, so broadcasting fails.
```

Broadcasting allows NumPy to perform math operations on arrays with different shapes
when the dimensions are compatible.

When using add(), NumPy lines up the arrays from right to left and stretches smaller
arrays to match larger arrays when possible.

The resulting array usually takes the shape of the larger array after broadcasting.

## Grocery Problem

```python
# Grocery bags
# Order: vegetables, fruits, snacks, beverages

bag1 = np.array([2, 3, 4, 0])

bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
# Total items in each bag

bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
# Total items in each category

totals = np.add(bag1, bag2)

print(f"There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
# Ali's door knocks
ali = np.array([5, 4, 6, 4, 2])

print(ali)
```

```python
# Ben's door knocks
# Monday = 0 because he missed that day

ben = np.array([0, 5, 5, 3, 4])

print(ben)
```

```python
# Carla's door knocks
# Wednesday = 0 because she was sick

carla = np.array([3, 6, 0, 3, 4])

print(carla)
```

```python
# Combine all values into one array

all_knocks = np.array([ali, ben, carla])

print(all_knocks)
```

```python
# Total doors knocked by each person

ali_total = np.sum(ali)
ben_total = np.sum(ben)
carla_total = np.sum(carla)

group_total = np.sum(all_knocks)

print(f"Ali knocked on {ali_total} doors.")
print(f"Ben knocked on {ben_total} doors.")
print(f"Carla knocked on {carla_total} doors.")
print(f"The group knocked on {group_total} doors total.")
```

```python
# Rolling total of estimated votes

daily_totals = np.sum(all_knocks, axis=0)

rolling_total = np.cumsum(daily_totals)

estimated_votes = rolling_total * 0.5

print(estimated_votes)
```
